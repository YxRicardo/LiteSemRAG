# SemCor Noun-Only Demo

This notebook focuses on noun annotations only.


```python
from itertools import islice
from pathlib import Path

import pandas as pd

from prepare_semcor import (
    get_sentence_record_by_index,
    iter_noun_annotation_records,
    iter_noun_sentence_records,
    load_semcor_stats,
)

BASE_DIR = Path("data/semcor")
```

```python
stats = load_semcor_stats(BASE_DIR)
pd.Series(stats)
```

```python
noun_sentence_sample = next(iter(iter_noun_sentence_records(BASE_DIR, limit=1)))
print(noun_sentence_sample["sentence_id"])
print(noun_sentence_sample["text"])

pd.DataFrame(noun_sentence_sample["noun_annotations"])[
    [
        "annotation_index",
        "text",
        "tokens",
        "lemma",
        "pos",
        "sense_key",
        "synset_name",
        "synset_definition",
    ]
]
```

```python
def inspect_sentence_by_index(sentence_index: int) -> pd.DataFrame:
    sentence = get_sentence_record_by_index(sentence_index, BASE_DIR)
    noun_annotations = sentence["noun_annotations"]

    print(f"sentence_index: {sentence_index}")
    print(f"sentence_id: {sentence['sentence_id']}")
    print("sentence_text:")
    print(sentence["text"])
    print()
    print("noun surface tokens:")
    print([annotation["tokens"] for annotation in noun_annotations])

    return pd.DataFrame(noun_annotations)[
        [
            "annotation_index",
            "text",
            "tokens",
            "lemma",
            "pos",
            "sense_key",
            "synset_name",
            "synset_definition",
        ]
    ]


inspect_sentence_by_index(0)
```

```python
noun_preview = list(islice(iter_noun_annotation_records(BASE_DIR), 20))
pd.DataFrame(noun_preview)[
    [
        "sentence_id",
        "text",
        "tokens",
        "lemma",
        "pos",
        "sense_key",
        "synset_name",
    ]
]
```

```python
def find_nouns_by_lemma(target_lemma: str, limit: int = 10) -> pd.DataFrame:
    rows = []
    for record in iter_noun_annotation_records(BASE_DIR):
        if record.get("lemma") == target_lemma:
            rows.append(
                {
                    "sentence_id": record["sentence_id"],
                    "text": record["text"],
                    "tokens": record["tokens"],
                    "lemma": record["lemma"],
                    "synset_name": record["synset_name"],
                    "definition": record["synset_definition"],
                }
            )
        if len(rows) >= limit:
            break
    return pd.DataFrame(rows)


find_nouns_by_lemma("evidence", limit=10)
```
